```python
from langchain.chat_models import ChatOpenAI
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.prompts import PromptTemplate
from langchain.memory import ConversationBufferMemory
from langchain.chains import LLMChain
from dotenv import load_dotenv
import os
```

```python
load_dotenv()
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
```

```python
detected_response = {
    'labels': ['FMCG', 'CPG', 'Fruits and Vegetables', 'Grains and Cereals', 'Consumer Electronics Accessories', 'Medicine'],
    'counts': [0, 0, 12, 10, 5, 0],
    'image': 'base64-image-placeholder'
}
```

```python
total_items = sum(detected_response['counts'])
category_summary = "\n".join(
    [f"{label}: {count} item(s)" for label, count in zip(detected_response['labels'], detected_response['counts'])]
)
```

```python
summary_block=f"Total products detected: {total_items}. Category breakdown:{category_summary}. \nGenerate a clear and concise shelf report in no more than 100 words. \n Response should not contain any titles or headings."
```

```python
template="""
You are a Shelfbot, an AI retail assistant that generates smart summary or report of product shelf layouts and arrangements based on product category and their counts. 
This is the list of product category we are considering : 
['FMCG', 'CPG', 'Fruits and Vegetables', 'Grains and Cereals', 'Consumer Electronics Accessories', 'Medicine'].
Identify dominant and missing categories, suggest improvements, and summarize the overall layout for business stakeholders.

Shelf detection summary:
{summary_block}

"""
```

```python
prompt = PromptTemplate(
    input_variables=["summary_block"],
    template=template
)
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-1.5-flash", temperature=0.5)

memory = ConversationBufferMemory()

shelfbot_chain=LLMChain(llm=llm,prompt=prompt, memory=memory, verbose=False)
```

```python
result=shelfbot_chain.run({"summary_block": summary_block})
```
